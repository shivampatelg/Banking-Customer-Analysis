```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 17, 30, 45, 60, 100],
    labels=['0-17', '18-30', '31-45', '46-60', '61+']
)
```

```python
df['age_group'].value_counts().sort_index()
```

```python
df['join_year'] = df['joined_bank'].dt.year
```

```python
df['join_year'].value_counts().sort_index()
```

```python
import numpy as np
```

```python
df['loan_to_deposit_ratio'] = np.where(
    df['bank_deposits'] > 0,
    df['bank_loans'] / df['bank_deposits'],
    np.nan
)
```

```python
import matplotlib.pyplot as plt
```

```python
# 1. Total Deposits by Loyalty Tier by Bar

data = df.groupby('loyalty_classification')['bank_deposits'].sum()
data.plot(kind = "bar")
plt.title("Total Deposits by Loyalty Tier")
plt.ylabel('Total Deposits')
plt.xlabel('Loyalty Tier')
plt.show()
```

```python
# 2. Loan-to-deposit ratio by age group (bar)

g = df.groupby('age_group')
ratio = g['bank_loans'].sum() / g['bank_deposits'].sum()
ratio.plot(kind = 'bar')
plt.title('Loan-to-deposit ratio by Age Group')
plt.ylabel('Ratio')
plt.xlabel('Age Group')
plt.show()
```

```python
# 3. New customers by joining year (line)
df.groupby('join_year').size().plot(kind = 'line')
plt.title('New Customers by Joining Year')
plt.ylabel('New Customers')
plt.xlabel('Joining Year')
plt.show()
```

```python
# 4. Income vs credit card balance (scatter)
df.plot(kind = 'scatter',x ='estimated_income', y = 'credit_card_balance')
plt.title("Income vs Credit Card Balance")
plt.show()

```
